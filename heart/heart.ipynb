```python
df["Cholesterol"].value_counts()
```

```python
#to replace chlosterol value (0) with mean
ch_mean=df.loc[df["Cholesterol"]!=0,"Cholesterol"].mean()
ch_mean
```

```python
#replaced 0 with mean value

df["Cholesterol"]=df["Cholesterol"].replace(0,ch_mean)
```

```python
#here we do the same thing for restingbp as no person can have restingbp zero
rest_mean=df.loc[df["RestingBP"]!=0,"RestingBP"].mean()
rest_mean
df["RestingBP"]=df["RestingBP"].replace(0,rest_mean)
```

```python
def plotting(var,num):
    plt.subplot(2,2,num)
    sns.histplot(df[var],kde=True)

plotting("Age",1)
plotting("RestingBP",2)
plotting("Cholesterol",3)
plotting("MaxHR",4)
plt.tight_layout()
```
