```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("train.txt" ,sep=';' ,header=None,names=["text","Emotion"])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions=df['Emotion'].unique()
Emotion_numbers={}
i=0
for emotion in unique_emotions:
    Emotion_numbers[emotion]=i
    i+=1

df['Emotion']=df['Emotion'].map(Emotion_numbers)
```

```python
df
```

```python
df["text"]=df["text"].apply(lambda x:x.lower())
```

```python
import string

def remove_punctuation(text):
    return text.translate(str.maketrans("","",string.punctuation))
```

```python
df["text"]=df["text"].apply(remove_punctuation)
```

```python
def remove_numbers(text):
    new=""
    for i in text:
        if not i.isdigit():
            new+=i
    return new

df["text"]=df["text"].apply(remove_numbers)
```

```python
def remove_emojis(text):
    new=""
    for i in text:
        if i.isascii():
            new+=i
    return new
df["text"]=df["text"].apply(remove_emojis)
        
```

```python
#have to use nltk library to remove stopwords
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download("stopwords")
nltk.download("punkt_tab")
nltk.download("punkt")
```

```python
stop_words=set(stopwords.words("english"))
```

```python
stop_words
```

```python
def remove_stopwords(text):
    word=word_tokenize(text)
    cleaned=[]
    for i in word:
        if i not in stop_words:
            cleaned.append(i)
    return ' '.join(cleaned)
```

```python
df["text"]=df["text"].apply(remove_stopwords)
```

```python
df.loc[1]["text"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df["text"], df["Emotion"], test_size=0.33, random_state=42)
```

```python
df.shape
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer


bow=CountVectorizer(ngram_range=(1,3))
X_train_bow=bow.fit_transform(X_train) 
X_test_bow=bow.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score

naive_model=MultinomialNB()
naive_model.fit(X_train_bow,y_train)
```

```python
y_pred=naive_model.predict(X_test_bow)
accuracy=accuracy_score(y_test,y_pred)
```

```python
print("Accuracy of the model is: ",accuracy)
```

```python
tf=TfidfVectorizer()


X_train_tf=tf.fit_transform(X_train)
X_test_tf=tf.transform(X_test)
naive_model.fit(X_train_tf,y_train)

y_predict_tf=naive_model.predict(X_test_tf)
accuracy_tf=accuracy_score(y_test,y_predict_tf)

```

```python

from sklearn.linear_model import LogisticRegression
linear_model=LogisticRegression(max_iter=1000)
linear_model.fit(X_train_tf,y_train)
y_linear=linear_model.predict(X_test_tf)

```

```python
accuracy_linear=accuracy_score(y_test,y_linear)
```
